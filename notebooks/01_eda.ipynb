```python
#separtion de entrer X et des cyble Y
import pandas as pd 
from sklearn.model_selection import train_test_split
df=pd.read_csv("../data/creditcard_enriched.csv")
features=[f"V{i}" for i in range(1, 29)]+ ["Amount_log" ,"Est_Nuit"]
X=df[features] # les entree 
y=df["Class"]

#decoupage en annales/examen
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
print("train : ", X_train.shape, "|Test: " ,X_test.shape) 
```

```python
#logistic regression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score
# 1. creer le modele(avec assez d'iterations pour converger)
model=LogisticRegression(max_iter=1000)
# entrener sur les annales 
model.fit(X_train, y_train)
#tester sur les suject jamais vus
y_pred=model.predict(X_test)

#4buletin de notes
print(classification_report(y_test,y_pred))
print("AUC-ROC: ",round(roc_auc_score(y_test,y_pred), 4))
```
